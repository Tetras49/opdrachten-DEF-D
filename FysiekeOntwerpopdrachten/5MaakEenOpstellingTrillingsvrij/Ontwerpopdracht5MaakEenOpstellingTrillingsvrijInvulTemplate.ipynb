## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
# De gewenste amplituderatio eis is 1/3 absoluut gezien, dus we rekenen met -1/3
gegeven_ratio = -1/3 # gebruik deze om de formule om te schrijven
f=5 # Gegeven frequentie
omega=f*2*np.pi
omega_0 = np.sqrt(omega**2/4)
f_cutoff = omega_0/(2*np.pi)
print("De cutoff frequentie is ", f_cutoff)
```

```python
# Voeg hier je plot in. 
omega=np.logspace(-1,2,1000) # maak waarden voor de x as
Am_Av_ratio = np.abs(1/(1-(omega/omega_0)**2)) # absoluut omdat anders de log een error geeft

plt.figure()
plt.loglog(omega,Am_Av_ratio)

plt.scatter(omega_0,1/3, label = "Gegeven criterium") #dit is de eis
plt.xlabel("Frequentie $\omega$ (rad/s)" )
plt.ylabel("Ratio van amplituden $Am/Av$" )
plt.grid(True, which="both")
plt.legend()
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
delta_x=9.81/omega_0**2
print("Minimale delta x is ", delta_x*100, "cm")
print("Voor een veilige marge kiezen we dus 4.5 cm")
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

plt.figure()
m=np.linspace(0.1,1.5,100)
C=(m*9.81)/delta_x # bereken C via de formule van het college, gebruik delta x zoals berekend bij é

plt.plot(m,C, label="Verband voor $\Delta x$")
plt.plot(m,C/2, label="Verband voor $2*\Delta x$")
plt.plot(m,C/5, label="Verband voor $5*\Delta x$")
plt.xlabel("Massa $m$ (kg)")
plt.ylabel("Veerconstante $C$ (N/m)")
plt.legend()
plt.grid()
plt.show()
```
